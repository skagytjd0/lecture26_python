### **학습 내용 되돌아보기(4)**  

````python
class Student:
    
    def __init__(self, name, dept, mid, final):
        self.name = name
        self.dept = dept
        self.mid = mid
        self.final = final
    def __str__(self):
        return f'학과: {self.dept}, 이름: {self.name}, 중간: {self.mid}, 기말: {self.final}'
    def grade(self):
        avg = (self.mid + self.final) /2
        if avg >= 90:
            return '학점: A'
        elif avg >= 80:
            return '학점: B'
        elif avg >= 70:
            return '학점: C'
        elif avg >= 60:
            return '학점: D'
        else:
            return '학점: F'
        
st2 = Student('김경철', '기계학과', 89, 90)
print(st2.__str__())
print(st2.grade())
````

### **학습 내용 되돌아보기(5)**  

````python
class Car:
    def __init__(self, name, year, color):
        self.name = name
        self.year = year
        self.color = color
    def __str__(self):
        return f'자동차 회사: {self.name}, 년식: {self.year}, 색상: {self.color}'
    def __eq__(self, other):
        # 파이썬에서는 조건식 자체를 반환 >> 어차피 True 또는 False로 값을 만들어냄
        return self.name == other.name and self.year == other.year and self.color == other.color
    
mycar = Car('현대', 2020, '검정')
yourcar = Car('기아', 2021, '백색')
print(mycar)
print(yourcar)
print(mycar == yourcar)
````
