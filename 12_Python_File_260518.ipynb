```python
filename = './sample.txt'

with open(filename, encoding='utf-8') as f_read:
    content = f_read.read()
print(content)
```

```python
with open(filename,encoding='utf-8') as f_read:
    lines = f_read.readlines()

for line in lines:
    print(line.strip())
```

```python
with open(filename,encoding='utf-8') as f_read:
    while True:
        lines = f_read.readline()
        if lines: print(lines.strip())
        else: break
```

```python
fruits = ['사과', '토마토', '바나나']

outfile = 'fruits.txt'
with open(outfile, mode='w', encoding='utf-8') as f_write:
    f_write.write('\n'.join(fruits))

with open(outfile, mode='a', encoding='utf-8') as f_write:
    f_write.write('\n'.join(fruits))
```

```python
with open(filename, mode='rb') as f_read:
    content = f_read.read()
print(content)
print(content.decode('utf-8'))
```

### 학습 내용 되돌아보기(2)

```python
f = open('pyzenex.txt', 'wt')
f.write('파이썬 철학\n')
f.write('아름다움이 추한 것보다 낫다.\n')
f.write('명시적인 것이 암묵적인 것보다 낫다.\n')
f.write('단순함이 복잡한 것보다 낫다.\n')

f = open('pyzenex.txt')
for line in f.readlines():
    print(line, end='')
f.close()

f = open('pyzenex.txt')
for line in f:
    print(line, end='')
f.close()

f = open('pyzenex.txt')
while True:
    line = f.readline()
    if not line: break
    print(line, end='')
f.close()
```

```python
cp = open('cp.txt')

# enumerate() : 번호 + 값 반환하는 파이썬 내장함수
for i, line in enumerate(cp):
    print(f'{i+1}학년 {line.strip()}')

cp.close()
```
